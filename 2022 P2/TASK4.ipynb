```python
#task 4.1

import sqlite3
con = sqlite3.connect("library.db")

#create book table
con.execute("CREATE TABLE IF NOT EXISTS Book (BookID INT PRIMARY KEY, Title TEXT, Genre TEXT);")
con.commit()


#create member table
con.execute("CREATE TABLE IF NOT EXISTS Member (MemberNumber INT PRIMARY KEY, FamilyName TEXT, GivenName TEXT);")
con.commit()


#create loan table
con.execute('''CREATE TABLE IF NOT EXISTS Loan (LoanID INT PRIMARY KEY, MemberNumber INT, BookID INT, DateLoaned TEXT, Returned TEXT,
                FOREIGN KEY (MemberNumber) REFERENCES Member(MemberNumber), 
                FOREIGN KEY (BookID) REFERENCES Book(BookID))''')
con.commit()

con.close()
```

```python
#task 4.2

#insert into book table
def insertbook():
    con = sqlite3.connect("library.db")
    with open("BOOK.txt", "r") as f:
        allLines = f.readlines()
        for line in allLines:
            bookid, title, genre = line.strip().split(",")
            con.execute("INSERT INTO book VALUES (?, ?, ?);", (bookid, title, genre))
            con.commit()
    con.close()

#insert into member table
def insertmember():
    con = sqlite3.connect("library.db")
    with open("MEMBER.txt", "r") as f:
        allLines = f.readlines()
        for line in allLines:
            memid, famname, name = line.strip().split(",")
            con.execute("INSERT INTO Member VALUES (?, ?, ?);", (memid, famname, name))
            con.commit()
    con.close()        

#insert into loan table
def insertloan():
    con = sqlite3.connect("library.db")
    with open("LOAN.txt", "r") as f:
        allLines = f.readlines()
        for line in allLines:
            loanid, memnum, bookid, date, returned = line.strip().split(",")
            con.execute("INSERT INTO Loan VALUES (?, ?, ?, ?, ?);", (loanid, memnum, bookid, date, returned))
            con.commit()
    con.close()
```
