```python
#task 2.1
import random 

def task2_1():
    global intlist
    intlist = []
    for i in range(100):
        intlist.append(random.randint(1, 100))
    return intlist

print(task2_1())
```

```python
#task 2.2

def bubblesort(arr):
    for i in range(0, len(arr)-1):
        for j in range(0, len(arr)-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]

    return arr

def bubble_sortlist():
    to_sort = task2_1()
    sortedlist = bubblesort(to_sort)
    return sortedlist

print(bubble_sortlist())

```

```python
#task 2.3

def mergesort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr)//2
    left = mergesort(arr[:mid])
    right = mergesort(arr[mid:])
    result, i, j = [], 0, 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    return result + left[i:] + right[j:]

def merge_sortlist():
    to_sort = task2_1()
    sortedlist = mergesort(to_sort)
    return sortedlist


while True:
    choice = input("Select which sorting algorithm you want to use(bubble/merge): ")
    
    if choice.lower() == "bubble":
        print(bubble_sortlist())
        break
    elif choice.lower() == "merge":
        print(merge_sortlist())
        break
    else:
        print("Invalid choice. Please choose correctly.")

```

```python
#task 2.4

def recBinarySearch(low, high, searchkey, arr):
    if low > high:
        return -1
    else:
        mid = (high+low)//2
        if searchkey == arr[mid]:
            return mid
        elif searchkey > arr[mid]:
            return recBinarySearch(mid+1, high, searchkey, arr)
        else:
            return recBinarySearch(low, mid-1, searchkey, arr)

def searchlist():
    arr = bubblesort(task2_1())
    value = int(input("Enter the value you want to search for: "))
    result = recBinarySearch(0, len(arr), value, arr)
    if result == -1:
        return "Not found"
    else:
        return result 
    
print(searchlist())
print(searchlist())
```
