```python
#task 1.1

def encrypt(character):
    if character.isalpha():
        value = ord(character)    #find the ASCII value of the character
        final_val = value + 10      #add 10 to the ASCII value
        if character.isupper() and final_val > ord("Z"):
            final_val = 65 + (final_val-ord("Z")-1)
        if character.islower() and final_val > ord("z"):
            final_val = 97 + (final_val-ord("z")-1)

        return chr(final_val)
    
    elif character.isspace():
        return "!"
        
    else:
        return -1
```

```python
#task 1.2

for i in range(4):
    char = input("Enter a single character: ")
    result = encrypt(char)
    if result == -1:
        print("Invalid input. Character has to be either letter or space.")
    else:
        print(result)
```

```python
#task 1.3

with open("DATATOENCRYPT.txt", "r") as f:
    allLines = f.readlines()
    for line in allLines:
        data = line.strip()

encrypted = ""
for i in data:
    result = encrypt(i)
    if result != -1:
        encrypted += result

with open("ENCRYPTEDMESSAGE.txt", "w") as f:
    f.write(encrypted)
```
