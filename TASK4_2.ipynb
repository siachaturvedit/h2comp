```python
#task 4.2
from datetime import datetime

class Person:
    def __init__(self, full_name, dob):
        self.full_name = full_name
        self.date_of_birth = datetime.strptime(dob, "%Y-%m-%d") #obtain the date string of date of birth, as year-month-date

    def is_adult(self):
        this_year = datetime.now().year #obtain the current year
        dob_year = self.date_of_birth.year #extract the year of birth
        if this_year - dob_year > 18:
            return True #return True if difference is greater than 18, otherwise return False
        return False
    
        
    def screen_name(self):
        screenname = ""
        for i in fullname:
            if i.isalpha():
                screenname += i   #add the letter into screenname if it is a character, effectively getting rid of spaces or punctuation
        screenname = screenname + str(self.date_of_birth.month) + str(self.date_of_birth.day) #add the birth month and birth day at the back
        return screenname
        


class Staff(Person):
    def __init__(self, full_name, dob):
        super().__init__(full_name, dob) #inherit attributes from parent class, Person

    def is_adult(self):
        return True #always return True

    def screen_name(self):
        screenname = ""
        for i in fullname:
            if i.isalpha():
                screenname += i
        screenname = screenname + str(self.date_of_birth.month) + str(self.date_of_birth.day) + "Staff"  #add staff to the back of the screenname
        return screenname



class Student(Person):
    def __init__(self, full_name, dob):
        super().__init__(full_name, dob)  #inherit attributes from parent class, Person

    def is_adult(self): 
        return False    #always return False
        
```

```python
#read from file, create instance, and insert info into school.db
import sqlite3
con = sqlite3.connect("school.db")

con.execute("CREATE TABLE IF NOT EXISTS People (PersonID INTEGER PRIMARY KEY AUTOINCREMENT, FullName TEXT, DateOfBirth TEXT, ScreenName TEXT, IsAdult INTEGER);")
con.commit()
con.close()

with open("people.txt", "r") as f:
    allLines = f.readlines()
    for line in allLines:
        full_name, dob, classtype = line.strip().split(",")
        if classtype == "Staff":
            person1 = Staff(full_name, dob)
        elif classtype == "Student":
            person1 = Student(full_name, dob)
        elif classtype == "Person":
            person1 = Person(full_name, dob)

        con.execute("INSERT INTO People (FullName, DateOfBirth, ScreenName, IsAdult) VALUES (?,?,?,?)", (full_name, dob, person1.screen_name(), person1.is_adult()))
        con.commit()

con.close()
```
