```python
#Task 3.1

def isValid(row):
    digits = row.split(" ")
    if len(digits) == 9:
        for i in digits:
            if not (1<=int(i)<=9):
                return False
    else:
        return False
    return True


def inputPuzzle():
    list_2d = [] 
    for i in range(9):
        lists_of_digits = input("Enter a list of digits: ")
        if isValid(lists_of_digits):
            digits = lists_of_digits.split(" ")
            list_2d.append(digits)
        else:
            list_2d.append([])

    return list_2d

example1 = inputPuzzle()
example2 = inputPuzzle()
```

```python
#task 3.2 
#write a function that checks each row of the puzzle to make sure it is completed horizontally
#return -1 if completed, otherwise return row number

def checkrows(example):
    check = []
    wrongrow = -1

    for i in range(9):
        for j in example[i]:
            check.append(j)
        checkset = set(check)
        if len(checkset) < 9:
            wrongrow = i + 1
        check = []

    return wrongrow


print(checkrows(example1))
print(checkrows(example2))
```

```python
#task 3.3

def checkcolumns(example):
    check = []
    wrongcolumn = -1

    for i in range(9):
        for j in range(9):
            check.append(example[j][i])
        checkset = set(check)
        if len(checkset) < 9:
            wrongcolumn = i + 1
        check = []

    return wrongcolumn

print(checkcolumns(example1))
print(checkcolumns(example2))
```
