```python
#task 2.1

class QueueItem:
    def __init__(self, ID, quantity):
        self.id = ID
        self.qty = quantity

    def getID(self):
        return self.id

    def getQuantity(self):
        return self.qty

    def to_str(self):
        return f"Item with ID {self.id} has quantity {self.qty}"

        
```

```python
#task 2.2, 2.3, 2.4, 2.5

class Queue:
    def __init__(self):
        self.queueList = [None] * 4     #initialise 4 null objects to queueList
        self.headPointer = 0
        self.tailPointer = 0
        self.numberItems = 0

    def isFull(self):
        return self.numberItems == 4

    def isEmpty(self):
        return self.numberItems == 0

    def enqueue(self, item):
        if self.isFull():
            print("Queue full. Cannot insert data.")
            return False
        else:
            self.queueList[self.tailPointer] = item
            self.tailPointer = (self.tailPointer + 1) % len(self.queueList) #wraps around the list when it reaches full capacity
            self.numberItems += 1
            return True

    def dequeue(self):
        if self.isEmpty():
            return False
        
        temp = self.queueList[self.headPointer]
        self.headPointer = (self.headPointer + 1) % len(self.queueList)
        self.numberItems -= 1
        return temp

    def printQueue(self):
        for i in self.queueList:
            if i == None:
                print("Null")
            else:
                print(i.getID(), ",", i.getQuantity())
        return
```

```python
#task 2.3

myQ3 = Queue()

for i in range(5):
    ID = input("Enter the ID of the item: ")
    qty = int(input("Enter the quantity of the item: "))
    new_item = QueueItem(ID, qty)
    if myQ3.enqueue(new_item):
        print(f"Item with id {ID} successfully added into queue")
        print(new_item.to_str())
    else:
        print(f"Queue full. Item with id {ID} not added into queue") 
        print(new_item.to_str())

```

```python
#Task 2.4

for j in range(5):
    item = myQ3.dequeue()
    if item != False:
        print(item.getID(), ",", item.getQuantity())
    else:
        print("Queue empty")
```

```python
#Task 2.5

myQ = Queue()

for i in range(4):
    ID = input("Enter the ID of the item: ")
    qty = int(input("Enter the quantity of the item: "))
    new_item = QueueItem(ID, qty)
    myQ.enqueue(new_item)

myQ.printQueue()
```
