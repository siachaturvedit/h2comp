```python
#Task 1.1

import sqlite3
con = sqlite3.connect("TESTDATABASE")

#create test table 
con.execute("CREATE TABLE IF NOT EXISTS Test (TestID TEXT PRIMARY KEY, MaxMarks INT)")
con.commit()


#create class table 
con.execute("CREATE TABLE IF NOT EXISTS Class (ClassName TEXT PRIMARY KEY, YearGroup INT)")
con.commit()


#create student table 
con.execute("CREATE TABLE IF NOT EXISTS Student (StudentID INT PRIMARY KEY, GivenName TEXT, FamilyName TEXT, ClassName TEXT, FOREIGN KEY (ClassName) REFERENCES Class(ClassName))")
con.commit()


#create student_test table 
con.execute("CREATE TABLE IF NOT EXISTS Student_Test (StudentTestID INT PRIMARY KEY, StudentID INT, TestID TEXT, Mark INT, DateTest TEXT, FOREIGN KEY (StudentID) REFERENCES Student(StudentID), FOREIGN KEY (TestID) REFERENCES Test(TestID))")
con.commit()

con.close()
```
