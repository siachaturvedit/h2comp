```python
#task 4.1
def read_file(filename):
    data = []
    with open(filename, "r") as f:
        allLines = f.readlines()
        for line in allLines:
            data.append(line.strip())
    return data

 
def compressor(colours):
    total = 0
    index = 0
    counter = 0
    string1 = ""
    
    while True:
        first = colours[index]
        string1 += str(first)
        counter += 1
        total += 1
        index += 1

        if total == len(colours):
            string1 += str(counter)
            break
            
        while colours[index] == first:
            counter += 1
            total += 1
            index += 1
        
        string1 += str(counter)
        counter = 0

    return string1


def write_file(comp_data):
    with open("compressedimage.txt", "w") as f:
        f.write(comp_data)

write_file(compressor(read_file("imagefile.txt")))
```

```python
#task 4.2

def read_compressed_file(filename):
    with open(filename, "r") as f:
        allLines = f.readlines()
        for line in allLines:
            data = line.strip()

    return data

def decompressor(data):
    datalist = []
    str1 = ""
    
    for i in range(0, len(data), 4):
        datalist.append(data[i:i+4])
    for j in datalist:
        for k in range(int(str(j)[3])):
            str1 += str(j[:3])
            str1 += "\n"

    return str1

def write_decompressed_file(decomp_data):
    with open("decompressedimage.txt", "w") as f:
        f.write(decomp_data)

write_decompressed_file(decompressor(read_compressed_file("compressedimage.txt")))

while True:
    print("Menu: ")
    print("1. Compress image\n2. Decompress image\n3. Quit")
    choice = int(input("Choose an option: "))

    if choice == 1:
        filename = input("Enter a file name to compress: ")
        write_file(compressor(read_file(filename)))

    elif choice == 2:
        filename = input("Enter a file name to decompress: ")
        write_decompressed_file(decompressor(read_compressed_file(filename)))

    else:
        break
```
