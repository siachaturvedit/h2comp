```python
#task 3.1

def task3_1(quantity_of_data):
    approved_units = ["KB", "MB", "GB", "TB"]
    if quantity_of_data[-2:] in approved_units:
        if quantity_of_data[:-2].isdigit() and int(quantity_of_data[:-2]) >= 0:
            if int(quantity_of_data[:-2]) == 0:
                print(0)
                return
            elif quantity_of_data[-2:] == "KB":
                print(int(quantity_of_data[:-2])*(10**3))
                return
            elif quantity_of_data[-2:] == "MB":
                print(int(quantity_of_data[:-2])*(10**6))
                return
            elif quantity_of_data[-2:] == "GB":
                print(int(quantity_of_data[:-2])*(10**9))
                return
            elif quantity_of_data[-2:] == "TB":
                print(int(quantity_of_data[:-2])*(10**12))
                return
    else:
        print("invalid data")
        return
            
task3_1("8KB")
task3_1("20MB")
task3_1("4GB")
task3_1("3TB")
```

```python
#task 3.2

def task3_2(quantity_of_data):
    approved_units = {"KB": 10**3, "MB": 10**6, "GB": 10**9, "TB": 10**12, "KiB": 2**10, "MiB": 2**20, "GiB": 2**30, "TiB": 2**40}
    for unit in approved_units:
        if quantity_of_data.endswith(unit):
            num = quantity_of_data[:-len(unit)]
            if num.isdigit() and int(num) >= 0:
                result = int(num) * approved_units[unit]
                print(result)
                return result
    else:
        print("invalid data")
        return False

task3_2("8KB")
task3_2("20MB")
task3_2("4GB")
task3_2("3TB")
task3_2("8KiB")
task3_2("512MiB")
task3_2("4GiB")
task3_2("3TiB")
```

```python
#task 3.3

def task3_3(quantity_of_data, target_unit):
    approved_units = {"KB": 10**3, "MB": 10**6, "GB": 10**9, "TB": 10**12, "KiB": 2**10, "MiB": 2**20, "GiB": 2**30, "TiB": 2**40}
    if target_unit in approved_units.keys():
        numBytes = task3_2(quantity_of_data)
        if numBytes != False:
            divisor = approved_units[target_unit]
            final = round((numBytes/divisor), 1)
            return final
    else:
        return "invalid data"

task3_3("512MiB", "GiB")
task3_3("512MB", "GB")
task3_3("5TB", "TiB")
```
