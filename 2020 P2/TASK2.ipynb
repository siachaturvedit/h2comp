```python
#task 2.1
import random

def task2_1(filename, quantity, maximum):
    nums = []
    for i in range(quantity):
        nums.append(random.randint(0, maximum))

    with open(filename, "w") as f:
        for j in nums:
            f.write(f"{str(j)}\n")
    
    
task2_1("randomnumbers.txt", 1000, 5000)
```

```python
#task 2.2

def task2_2(list_of_integers):
    if len(list_of_integers) <= 1:
        return list_of_integers #set base case

    mid = len(list_of_integers) // 2
    left = task2_2(list_of_integers[:mid])
    right = task2_2(list_of_integers[mid:])
    result, i, j = [], 0, 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    return result + left[i:] + right[j:]

intList = [56, 25, 4, 98, 0, 18, 4, 5, 7, 0]
print(task2_2(intList))

print(task2_2(intList) == [0, 0, 4, 4, 5, 7, 18, 25, 56, 98])
```

```python
#task 2.3

def task2_3(filename_in, filename_out):
    ints = []
    with open(filename_in, "r") as f:
        allLines = f.readlines()
        for line in allLines:
            ints.append(int(line))

    sorted_list = task2_2(ints)
    
    with open(filename_out, "w") as f:
        for num in sorted_list:
            f.write(f"{num}\n")

task2_3("randomnumbers.txt", "sortednumbers.txt")
```
