```python
#task 2.1

def task2_1(filename):
    intlist = []
    with open(filename, "r") as f:
        allLines = f.readlines()
        for line in allLines:
            intlist.append(int(line.strip()))

    return intlist 

result = task2_1('TEN.txt')
print(result)
print(len(result))
```

```python
#task 2.2

def task2_2(list_of_integers):
    for i in range(1, len(list_of_integers)):
        while i>0 and list_of_integers[i-1] > list_of_integers[i]:
            list_of_integers[i-1], list_of_integers[i] = list_of_integers[i], list_of_integers[i-1]
            i = i-1

    return list_of_integers

print(task2_2(task2_1('TEN.txt')))
```

```python
#task 2.3

def task2_3(list_of_integers):
    if len(list_of_integers) <= 1:
        return list_of_integers
    pivot = list_of_integers[0]
    left = [x for x in list_of_integers[1:] if x<=pivot]
    right = [x for x in list_of_integers[1:] if x>pivot]
    left = task2_3(left)
    right = task2_3(right)
    return left + [pivot] + right

print(task2_3(task2_1('TEN.txt')))
```

```python
#task 2.4

import timeit # import the built-in library timeit

def timefunction(filename, taskname):
    intlist = task2_1(filename)
    time = timeit.timeit(lambda: taskname(intlist), number=1)
    return time


#compare "TEN.txt"
print("Compare with list of 10 integers: ")
instime = timefunction("TEN.txt", task2_2)
quicktime = timefunction("TEN.txt", task2_3)
print(f"Insertion sort: {instime}\nQuicksort: {quicktime}\n")


#compare "HUNDRED.txt"
print("Compare with list of 100 integers: ")
instime = timefunction("HUNDRED.txt", task2_2)
quicktime = timefunction("HUNDRED.txt", task2_3)
print(f"Insertion sort: {instime}\nQuicksort: {quicktime}\n")


#compare "THOUSAND.txt"
print("Compare with list of 1000 integers: ")
instime = timefunction("THOUSAND.txt", task2_2)
quicktime = timefunction("THOUSAND.txt", task2_3)
print(f"Insertion sort: {instime}\nQuicksort: {quicktime}")


print('''Insertion sort is quicker than quicksort for small arrays, such as in the case of a list with only 10 integers. 
        However, since insertion sort has the worst-case order of growth of O(n^2), while insertion sort has average and 
        best case order of growth of O(nlogn), quicksort is much faster in sorting larger sets of data, as shown for the lists 
        with a hundred and thousand integers. ''')
```
