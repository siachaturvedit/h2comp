```python
#task 1.1

def task1_1(input_value):
    inputstr = str(input_value)
    if inputstr.isdigit() and len(inputstr) == 14:  #check the parameter is either int or string containing int, and that it has 14 digits
        checkstr = ""
        for i in range(len(inputstr)):
            if i%2 == 1:
                checkstr += str(int(inputstr[i])*2) #multiple the numbers at odd positions by two and add into the string
            else:
                checkstr += str(inputstr[i])

        checksum = 0
        for j in checkstr:
            checksum += int(j) #sum up all the numbers

        rem = checksum % 10
        if rem == 0:
            checkdigit = 0
        else:
            checkdigit = 10 - rem
        
        return checkdigit
            
    else:
        return -1
        
```

```python
#task 1.2

print(task1_1("18573058693856") == 4)
print(task1_1(93728573697832) == 6)
print(task1_1("qwertyuiopasdf") == -1)
print(task1_1(12345) == -1)
```

```python
#task 1.3

def task1_3(input_value):
    inputstr = str(input_value)       #declare as string for slicing later on and to check whether it is made up of digits
    if inputstr.isdigit() and len(inputstr) == 15:        #check whether input is all digits, and there are exactly 15 digits
        checknum = inputstr[-1]          #check digit is the last digit of the 15-digit string 
        first14 = inputstr[:-1]          #slice the first 14 digits to run in task1_1(first14) to get the check digit
        
        if task1_1(first14) == int(checknum):
            return True
        
    else:
        return False

print(task1_3("185730586938564"))
print(task1_3("937285736978326"))
print(not task1_3("qwertyuiopasdf"))
print(not task1_3(12345))
```
