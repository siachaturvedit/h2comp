```python
#task 3.1

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, integer_value):
        newnode = Node(integer_value)
        newnode.next = self.head
        self.head = newnode

    
    def delete(self, integer_value):
        if self.head is None:      #check whether the linked list is empty
            print("Linked list does not exist.")
            return
        else:
            found = False
            current = self.head
            previous = None     #set previous pointer to keep track of previous node

            while not found:
                if current.data == integer_value:
                    found = True       #look for the node iteratively
                else:
                    previous = current
                    current = current.next

            if previous == None:     #if the item is the first item in the list 
                self.head = current.next      #the first item will be the one after the one that was deleted
            else:      #otherwise, if the item was in the middle of the linked list
                previous.next = current.next     #the node before the item will point to the one after it
            

    def search(self, integer_value):
        if self.head is None:      #check whether the linked list is empty
            print("Linked list does not exist.")
            return
        else:
            current = self.head
            found = False
            while current != None and not found:
                if current.data == integer_value:
                    found = True 
                else:
                    current = current.next
        
            return found 
            
            

    def count(self):
        if self.head is None:      #check whether the linked list is empty
            print("Linked list does not exist.")
            return
        else:
            current = self.head
            count = 0
            while current != None:
                count += 1
                current = current.next


    
    def to_String(self):
        if self.head is None:      #check whether the linked list is empty
            print("Linked list does not exist.")
            return
        else:
            liststr = "["
            current = self.head
            while current != None:
                liststr += f"{str(current.data)}, "
                current = current.next

        liststr = liststr[:-2] + "]"
        
        return liststr


myLinkedList = LinkedList()
with open("Task3data.txt", "r") as f:
    allLines = f.readlines()
    for line in allLines:
        myLinkedList.insert(line.strip())

print(myLinkedList.to_String())
```

```python
#task 3.2

class SortedLinkedList(LinkedList):
    def __init__(self):
        super().__init__()

    def insert(self, data):
        newnode = Node(data)
        prev = None    #set a previous pointer to keep track of the previous node, so can insert node in the middle of other nodes later
        current = self.head

        if current is None or data < current.data:   #if the value being inserted is lesser than the rest, or the linked list is empty, insert it in the front - assign it as the head
            newnode.next = self.head 
            self.head = newnode
            return

        else:
            while current is not None and data >= current.data:   #iterate through the linked list until the new node is at its right place
                prev = current
                current = current.next    #increment both the previous and current nodes to find which values the data to insert falls between
     
            newnode.next = current   
            prev.next = newnode     #insert the new node in between the previous node with lower value, and the next node with higher value


myLinkedList = SortedLinkedList()

with open("Task3data.txt", "r") as f:
    allLines = f.readlines()
    for line in allLines:
        myLinkedList.insert(line.strip())

print(myLinkedList.to_String())
print(myLinkedList.search(94))
```

```python
#task 3.3

class Queue(LinkedList):
    def __init__(self):
        super().__init__()

    
    def enqueue(self, integer_value):
        newnode = Node(integer_value)
        current = self.head

        if self.head is None:
            self.head = newnode
            return
        
        while current.next is not None:
            current = current.next

        current.next = newnode
    

    def dequeue(self):
        if self.head is None:      #check whether the linked list is empty
            return None
        else:
            current = self.head
            self.head = self.head.next
            return current.data


myQ = Queue()

with open("Task3data.txt", "r") as f:
    allLines = f.readlines()
    for line in allLines:
        myQ.enqueue(line.strip())

for i in range(5):
    print(myQ.dequeue())
```
